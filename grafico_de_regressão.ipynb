```python
import pandas as pd

url = 'https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/refs/heads/main/student_exam_scores.csv'
df = pd.read_csv(url)

display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Verificar o nome exato das colunas no DataFrame
print("Colunas disponíveis:", df.columns.tolist())

# Criar o gráfico de dispersão
# Geralmente as colunas são 'hours_studied' e 'exam_score'
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score', color='blue', alpha=0.7)

plt.title('Relação entre Horas de Estudo e Nota da Prova')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota da Prova')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Calcular os coeficientes da regressão linear (y = ax + b)
x = df['hours_studied']
y = df['exam_score']
a, b = np.polyfit(x, y, 1)

# Criar a equação em formato de texto
equacao = f'y = {a:.2f}x + {b:.2f}'

# Configurar o gráfico
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score', color='blue', alpha=0.7, label='Dados reais')

# Plotar a linha de tendência
x_linha = np.linspace(x.min(), x.max(), 100)
y_linha = a * x_linha + b
plt.plot(x_linha, y_linha, color='red', linewidth=2, label=f'Linha de Tendência ({equacao})')

# Adicionar detalhes ao gráfico
plt.title('Relação entre Horas de Estudo e Nota da Prova com Linha de Tendência')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota da Prova')

# Posicionar o texto da equação no gráfico
plt.text(x.min() + 0.5, y.max() - 2, equacao, fontsize=12, color='red', weight='bold',
         bbox=dict(facecolor='white', alpha=0.8, edgecolor='red'))

plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
